#### 🚀 Exercício 1 (nível real de engenheiro)

Vamos montar um mini laboratório com:

- Ollama
- Python Notebook
- um prompt controlado
- validação de respostas

**Objetivo:**

Comparar:

- resposta do modelo
- resposta corrigida (ground truth)
- detecção de erro

```python
import requests
import json

OLLAMA_URL = "http://localhost:11434/api/generate"
MODEL = "llama3.1:8b"

def ask_llm(prompt):
    payload = {
        "model": MODEL,
        "prompt": prompt,
        "stream": False
    }

    response = requests.post(OLLAMA_URL, json=payload)
    return response.json()["response"]


# Pergunta propositalmente “enganosa”
prompt = """
Explique o que é LangGraph e sua origem.
Responda de forma técnica e objetiva.
"""

resposta = ask_llm(prompt)

print("\n===== RESPOSTA DO MODELO =====\n")
print(resposta)
```
